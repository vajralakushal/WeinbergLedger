```python
import pandas as pd
import random
import numpy as np
```

```python
df = pd.read_csv('Weinberg_Library_9_312_ascii.csv', keep_default_na=True)
```

```python
df = df.replace({np.nan: None})
```

```python
df.head()
```

```python
from db import connect, init_db
```

```python
connector = connect("library.db")
init_db(connector)
```

```python
id_set = set()
def random_id():
    num = random.randint(1, 10000)
    if num not in id_set:
        id_set.add(num)
        return num
    else:
        return random_id()
```

```python

for row in df.itertuples():
    #TODO: try/catch for not null fields
    id = random_id()
    owner = None if row.Owner is None else row.Owner
    borrower = None if row.Borrower is None else row.Borrower
    location = None if row.Shelf is None else row.Shelf
    title = None if row.Title is None else row.Title
    creator = None if row.Creator is None else row.Creator
    publisher = None if row.Publisher is None else row.Publisher
    series = None if row.Series is None else row.Series
    subject = None if row.Subject is None else row.Subject
    creation_date = None if row.CreationDate is None else row.CreationDate
    identifier = None if row.Identifier is None else row.Identifier
    
    data = (id, owner, borrower, location, title, creator, publisher, series, subject, creation_date, identifier)
    connector.execute(
    """
    INSERT INTO LIBRARY 
    (ID, OWNER, BORROWER, LOCATION, TITLE, CREATOR, PUBLISHER, SERIES, SUBJECT, CREATION_DATE, IDENTIFIER)
    VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
    """
    , data)
    connector.commit()
```
